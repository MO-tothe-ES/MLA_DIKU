```python
x_train = np.loadtxt('data/X_train.csv', delimiter=',')
y_train = np.loadtxt('data/y_train.csv')
display(x_train[:,0], y_train)
```

```python
def sigmoid(s):

    theta = 1 / (1+np.exp(-s))
    
    return theta

def grad_func(x_data, y_data, w, b):

    gradient = -np.mean(
        (y_data-sigmoid(w.T@x_data+b)) @ x_data
    )

    return gradient


def GD_minimizer(x_data, y_data, w, b, grad_func, learning_rate=0.1, tol=1e-8):

    wb = np.append(w,b)

    step_len = np.inf
    while step_len > tol:

        grad = grad_func(x_data,y_data,w,b)

        wb_plus = wb - learning_rate * grad
        
        step_len =  np.sqrt((wb_plus - wb).T @ (wb_plus - wb))

        wb = wb_plus

    return wb
```

```python
GD_minimizer(x_train,y_train)
```
